## I. Shorthand Operators

A shorthand is a shorter way to write an expression


| Operator  | Shorthand | Expression   |
|-----------|-----------|--------------|
| +=        | x += y    |  x = x + y   | 
| -=        | x -= y    |  x = x - y   | 
| *=        | x *= y    |  x = x * y   | 
| /=        | x /= y    |  x = x / y   | 
| %=        | x %= y    |  x = x % y   | 
| **=        | x **= y    |  x = x ** y   | 
| //=        | x //= y    |  x = x // y   | 


````python
# Example
x = 10
x += 1  # equivalent to x = x + 1
print(x)
````

### Hands-on Exercise 1:


a) Write a function named `swapCase` to swap cases of a given string. The function accepts a string as parameter and returns the resulted string.

Hint: use isupper() or islower() to check the case of each character

````python
def swapCase(caseCheck):
    result = ""
    for char in caseCheck:
        if char.isupper():
            result += char.lower()
        elif char.islower():
            result += char.upper()
    return result
````

````python
# Test your function
print(swapCase("COMP1007")) # comp1007
print(swapCase("Python Programming")) # pYTHON pROGRAMMING
````

b) Write a function named `letterCount` to return the number of occurrences of a letter in a given string. The function has two parameters: a given string and the letter to be counted in the given string. 

Note: The function should ignore the case of the letters


````python
def letterCount(mainBody, letter):
    count = 0
    for char in mainBody:
        if char == letter:
            count += 1
    return count 
````

````python
# Test your function
print(letterCount("Python Programming", "p"))  # 2
print(letterCount("Computer Science", "e"))  # 3
print(letterCount("Ascii Code", "c"))  # 2
````

## III. ASCII Code

Computer stores data in binary format. Every piece of information, including characters, numbers, and even program instructions, is stored as a sequence of 0’s and 1’s. For example, a lowercase ‘a’ is represented by 1100001 (97 in decimal) and a ‘b’ is encoded as 1100010 (98 in decimal). This encoding is used to identify a character’s ASCII code (American Standard Code for Information Interchange). Every character that appears on your keyboard has its own 7-bit ASCII sequence or code.

### Hands-on Exercise 2: Caesar Cipher

The Caesar Cipher technique is one of the earliest and simplest method of encryption technique. Each letter of a given text is replaced by a letter some fixed number of positions down the alphabet. For example with a shift of 1, A would be replaced by B, B would become C, and so on. Thus to cipher a given text we need an integer value, known as shift which indicates the number of position each letter of the text has been moved down. 

a) Write a function named `encrypt` to encrypt a given string by Caesar Cipher technique. The function accepts a text (string) in upper case letters and a shift value (integer between 0 - 25) as parameters and returns the encrypted text.

````python
def encrypt(string, s):
    newVal = ""
    for char in string:
        newVal += char + str(s)
    return newVal
````

````python
# Test function
text = "ABCD"
s = 3
print ("Text: " + text)
print ("Shift: " + str(s))
print ("Encrypted text: " + encrypt(text,s))  #DEFG
print()

text = "XYZ"
s = 1
print ("Text: " + text)
print ("Shift: " + str(s))
print ("Encrypted text: " + encrypt(text,s))  #YZA
print()

text = "PYTHON"
s = 5
print ("Text: " + text)
print ("Shift: " + str(s))
print ("Encrypted text: " + encrypt(text,s))  #UDYMTS
````

b) Write a function named `decrypt` to decrypt a given encrypted text. The function accepts a encrypted text (string) in upper case letters and a shift value (integer between 0 - 25) as parameters and returns the decrypted text.

````python
def decrypt(string1, s1):
    newVal1 = ""
    for char in string1:
        newVal1 += chr(ord(char) - s1)
    return newVal1
````

````python
# Test function
text = "DEFG"
s = 3
print ("Text: " + text)
print ("Shift: " + str(s))
print ("Decrypted text: " + decrypt(text,s))  #ABCD
print()

text = "YZA"
s = 1
print ("Text: " + text)
print ("Shift: " + str(s))
print ("Decrypted text: " + decrypt(text,s))  # XYZ
print()

text = "UDYMTS"
s = 5
print ("Text: " + text)
print ("Shift: " + str(s))
print ("Decrypted text: " + decrypt(text,s)) # PYTHON
````

## IV. Python Dictionary

Python dictionary is an unordered collection of items. Each item of a dictionary has a key-value pair.

You can define a dictionary by enclosing a comma-separated list of key-value pairs in curly braces ({}). A colon (:) separates each key from its associated value:

```
d = {
    <key>: <value>,
    <key>: <value>,
      .
      .
      .
    <key>: <value>
}
```

You can also create a dictionary with the built-in dict() function.

```
d = dict([
    (<key>, <value>),
    (<key>, <value),
      .
      .
      .
    (<key>, <value>)
])
```

````python
# Example

empty = {} # empty dictionary

score1 = {'Ada': 100, 'Bob': 58, 'Carl': 96}

score2 = dict([('Ada',100), ('Bob',58), ('Carl',96)]) # with dict() 

print(empty)
print(score1)
print(score2)
````

#### Accessing elements from a Dictionary

Use square brackets along with they key to get its value. 

````python
score = {'Ada': 100, 'Bob': 58, 'Carl': 96}

print(score['Bob'])
````

However, if the key is not in the dictionary, error occurs.

````python
score = {'Ada': 100, 'Bob': 58, 'Carl': 96}

print(score['Beth']) # Error
````

We can use an if statement to check whether a key is in the dictionary or not.

````python
score = {'Ada': 100, 'Bob': 58, 'Carl': 96}

key = 'Beth' 
if key in score:
    print(score[key])
else:
    print('No such key')
````

#### Looping through a Dictionary

````python
for key in score:
    print('key:', key)  # key
    print('value:', score[key]) # value
````

#### Updating and Adding elements to a Dictionary

If the key is already present, then the existing value gets updated. In case the key is not present, a new key-value pair is added to the dictionary.

````python
score = {'Ada': 100, 'Bob': 58, 'Carl': 96}
print(score)

score['Bob'] = 60  # update
print(score)

score['Ann'] = 70 # add a new key-value pair
print(score)
````

#### Removing elements from a Dictionary

````python
score = {'Ada': 100, 'Bob': 58, 'Carl': 96}
print(score)

del score['Bob'] # remove Bob
print(score)
````

### Hands on Exercise 3: Python Dictionary

a) Write a program to prompt user to enter a name and the corresponding price of a product, then store the name as key and price as value into a python dictionary . The program should ask the user whether he wants to enter the next product after each input and print the dictionary after the user ends the input.

Sample output:

```
Enter the product name:  apple
Enter the price:  2.5
Contine for next product [y/n]?  y
Enter the product name:  orange
Enter the price:  3.5
Contine for next product [y/n]?  y
Enter the product name:  banana
Enter the price:  4.2
Contine for next product [y/n]?  n
{'apple': 2.5, 'orange': 3.5, 'banana': 4.2}
```

````python
productList = {}
valid = True

while valid:
    productName = input("Enter a product name: ")
    productPrice = int(input("Enter the price: "))
    productList[productName] = productPrice
    
    buffer = input("Continue for next product [y/n]? ")
    
    if buffer == 'n':
        print(productList)
        valid = False  
    elif buffer != 'y':
        print("Invalid input, please try again :)")
````

b)Write a function called `cube` that accpets an integer n as parameter and returns a dictionary that contains integer 1 to n as key and the cube of the integer as value.

````python
def cube(n):
    dictionary1 = {}
    i = 0
    while i < n:
        i += 1
        dictionary1[i] = i**3 # add a new key-value pair
    return dictionary1
````

````python
# Test your function
print(cube(5)) # {1: 1, 2: 8, 3: 27, 4: 64, 5: 125}
````

c) Use the function in part(b) to generate the cube of 1 to 20.

````python
print(cube(20))
````

#####　Accessing the values in List of Dictionaries

````python
pList = [{'name': 'Alice', 'age': 18, 'gender': 'F'},  # index 0
         {'name': 'Bob', 'age': 21, 'gender': 'M'},    # index 1
         {'name': 'Cindy', 'age': 25, 'gender': 'F'}]  # index 2

print(pList[0])
print(pList[0]['name'])
print(pList[0]['age'])
print(pList[0]['gender'])
print()
print(pList[1])
print(pList[1]['name'])
print(pList[1]['age'])
print(pList[1]['gender'])
print()
print(pList[2])
print(pList[2]['name'])
print(pList[2]['age'])
print(pList[2]['gender'])
````

##### Looping thourgh List of Dictionaries

````python
pList = [{'name': 'Alice', 'age': 18, 'gender': 'F'},  # index 0
         {'name': 'Bob', 'age': 21, 'gender': 'M'},    # index 1
         {'name': 'Cindy', 'age': 25, 'gender': 'F'}]  # index 2

for d in pList:
    for k in d:
        print("key:", k)
        print("value:", d[k])     
````

````python
# Alternatively
pList = [{'name': 'Alice', 'age': 18, 'gender': 'F'},  # index 0
         {'name': 'Bob', 'age': 21, 'gender': 'M'},    # index 1
         {'name': 'Cindy', 'age': 25, 'gender': 'F'}]  # index 2

for i in range(len(pList)): 
    for k in pList[i]:
        print("key:", k)
        print("value:", pList[i][k])     
````

### Hands-on Exercise 4: Gradebook

a) Create three dictionaries: a, b and c. Give each dictionary the keys "name", "homework", "quiz", and "exam". Assign values to the keys as follow:

| Dictionary | name | homework | quiz | exam |
|---|---|---|---|---|
| a | Ada | 92.0 | 85.0 | 70.0 |
| b | Bob | 100.0 | 95.0  | 89.0 |
| c | Cindy | 0.0 | 50.0 | 45.0 |



````python
#dicionaries
a={'name':'Ada', 'homework':92.0, 'quiz':85.0, 'exam':70.0}
b={'name':'Bob', 'homework':100.0, 'quiz':95.0, 'exam':89.0}
c={'name':'Cindy', 'homework':0.0, 'quiz':50.0, 'exam':45.0}
````

b) Create a list called `students` that contains the dictionaries in part(a)

````python
students = []
students.append(a) 
students.append(b)
students.append(c)
````

c) For each student in the `students` list, print out all the key-value pairs.

Sample output:
```
name Ada
homework 92.0
quiz 85.0
exam 70.0
name Bob
homework 100.0
quiz 95.0
exam 89.0
name Cindy
homework 0.0
quiz 50.0
exam 45.0
```

````python
for student in students:
    for key, value in student.items():
        print(f"{key} {value}")
````

c) Suppose the weight of homework is 10%, quiz is 30% and exam is 60%. Add a key "average" to each student in the `students` list and assign the average score to it. Print the resulted `students` list.

Sample output:

```
[{'name': 'Ada', 'homework': 92.0, 'quiz': 85.0, 'exam': 70.0, 'average': 76.7}, {'name': 'Bob', 'homework': 100.0, 'quiz': 95.0, 'exam': 89.0, 'average': 91.9}, {'name': 'Cindy', 'homework': 0.0, 'quiz': 50.0, 'exam': 45.0, 'average': 42.0}]
```

````python
for d in students:
    # Calculate weighted average: 10% homework + 30% quiz + 60% exam
    avg = d['homework'] * 0.10 + d['quiz'] * 0.30 + d['exam'] * 0.60
    
    # Store the rounded average in the dictionary
    d['average'] = round(avg, 1)

# Print the updated list of dictionaries
print(students)
````
